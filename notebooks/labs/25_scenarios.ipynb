# Lab 25: Costing a Budget Cut

This lab reproduces Chapter 25's four costings of a 15 per cent budget cut on the teaching response curves: the average-ROAS shortcut, an equal cut, the targeted cut with reallocation within 40 per cent of each channel's budget, and the targeted cut where no channel may rise.
It accompanies Chapter 25 of the book Marketing Mix Modelling.

## The curves

The teaching curves are the ones a model would have if it had recovered the planning values exactly, the ones Chapters 24 and 25 use.

```python
import ex_scenarios

curves = ex_scenarios.teaching_curves()
revenue = ex_scenarios.annual_revenue(curves, curves.current_money)
spend = curves.total_budget * 52
print(f"current mix: {revenue / 1e6:.2f}m on {spend / 1e6:.1f}m")
```

## Four costings of the 15 per cent cut

Profit is half the revenue change plus the spend saved.
The shortcut is the only costing that shows a loss of profit.

```python
result = ex_scenarios.scenarios(curves)
for name, row in result.items():
    print(f"{name:<30}{row['revenue']:>+8.2f}{row['profit']:>+8.2f}")
assert abs(result["targeted, within +/-40%"]["profit"] - 0.44) < 0.01
```
