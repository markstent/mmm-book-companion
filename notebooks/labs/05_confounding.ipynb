## Conditioning on a website visit (Table 5.2)

The people are built to the chapter's recipe, and advertising adds two points of conversion for everyone.
Among all visitors, the advertised group converts less, because adverts brought in casual visitors who would otherwise have stayed away.

```python
import ex_confounding

v = ex_confounding.visitors_table()
print(f"all visitors: {v['visitors_ad']:.1%} advertised vs {v['visitors_no_ad']:.1%} not")
assert round(v["visitors_ad"], 3) == 0.153 and round(v["visitors_no_ad"], 3) == 0.217
```

Holding intent fixed removes the artificial gap, and the two points reappear in both groups.

```python
print(f"high intent: {v['high_visitors_ad']:.1%} vs {v['high_visitors_no_ad']:.1%}")
print(f"low intent:  {v['low_visitors_ad']:.1%} vs {v['low_visitors_no_ad']:.1%}")
```

## A promotion riding on television (Table 5.1)

Over 156 simulated weeks a 20 per cent promotion launches in almost every television week.
Television's true effect is 85,000 a week and the promotion's is 150,000.
A regression without the promotion gives television most of the promotion's credit, and adding the promotion as a control returns it.

```python
w = ex_confounding.promotion_weeks()
print(f"promotion left out: {w['naive_tv']:,.0f}")
print(f"promotion included: {w['adjusted_tv']:,.0f}")
assert abs(w["adjusted_tv"] - 85_000) < 10_000
```
