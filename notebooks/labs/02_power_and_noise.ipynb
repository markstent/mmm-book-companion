This lab simulates the four-week campaign of Chapter 2, which adds 20,000 pounds a week against weekly noise of 20,000 pounds (standard deviation), and counts how often a one-sided test at 1.96 standard errors detects it.

## The chapter's campaign

Each simulated campaign draws four weeks of noise, adds the uplift, and averages them.
The average has a standard error of 20,000 / sqrt(4) = 10,000, so the true effect sits two standard errors from zero.

```python
import ex_power

power = ex_power.simulate_power(uplift=20_000, noise_sd=20_000, weeks=4)
print(f"simulated power: {power:.1%}")
print(f"normal arithmetic: {ex_power.exact_power(20_000, 20_000, 4):.1%}")
assert abs(power - 0.516) < 0.01
```
