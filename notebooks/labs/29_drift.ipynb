# Lab 29: Drift and the Run-of-Four Rule

This lab scores new weeks against a 90 per cent predictive interval and applies Chapter 29's rule: four consecutive weeks outside the interval on the same side.
A fitted model's posterior predictive draws would supply the interval in a real refresh.
To keep the lab instant, it uses a stand-in with the chapter's numbers, a forecast of 900,000 pounds a week with 20,000 pounds of noise, so no model is fitted.

## The interval and the rule

```python
import ex_drift

low, high = ex_drift.predictive_interval()
print(f"90% interval: {low:,.0f} to {high:,.0f}")
print(f"four misses in a row, fixed side: {0.05**4:.1e}")
```

## Steady weeks, then a shift

With no change in the business the rule stays quiet.
Then revenue falls by 50,000 pounds a week from week 13, and the rule fires when the fourth miss arrives.

```python
steady = ex_drift.new_weeks()
shifted = ex_drift.new_weeks(shift=-50_000.0)
print("steady:", ex_drift.first_alarm(steady, low, high))
print("shifted:", ex_drift.first_alarm(shifted, low, high))
assert ex_drift.first_alarm(steady, low, high) is None
assert ex_drift.first_alarm(shifted, low, high) is not None
```

## False alarms

Over many 52-week stretches with no change, the rule fires rarely, because independent misses are rare.
Residuals are autocorrelated in real data, so the true rate is higher.

```python
print(f"{ex_drift.false_alarm_rate():.2%}")
```
